### Configurations && imports

```python
# dependencies
import requests
from bs4 import BeautifulSoup
import csv
from urllib.parse import urljoin
```

```python
# fetching the web content
url = "https://quotes.toscrape.com/"
response = requests.get(url)
html_content = response.text

# parsing the html
soup = BeautifulSoup(html_content, 'html.parser')
```

### Data extraction

```python
scraped_quotes = []

for element in soup.find_all("div", class_="quote"):
    quote = element.find("span", class_="text").get_text(strip=True)
    author = element.find("small", class_="author").get_text(strip=True)

    tags = []
    for t in element.find_all("a", class_="tag"):
        tag_name = t.get_text(strip=True)
        tags.append(tag_name)

    
    quote_dict = {
        "quote": quote,
        "author": author,
        "tags": tags
    }
    scraped_quotes.append(quote_dict)
```

```python
print(scraped_quotes[0])
```

### Dataset creation

```python
fieldnames = ['quote', 'author', 'tags']
with open('quotes.csv', 'w', newline='', encoding='utf-8') as csvfile:
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(scraped_quotes)

print("Data saved to products.csv")
```
