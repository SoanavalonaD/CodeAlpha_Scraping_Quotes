### 1) Configurations && imports

```python
# dependencies
import requests
from bs4 import BeautifulSoup
import csv
from urllib.parse import urljoin
from urllib3.util.retry import Retry
from requests.adapters import HTTPAdapter
import urllib.robotparser as robotparser
```

### 2) Requests session with retries & polite header

```python
def create_session():
    s = requests.Session()
    s.headers.update({
        "User-Agent": "Mozilla/5.0 (X11; Ubuntu) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117 Safari/537.36 - learning-scraper"
    })
    # retires in case of errors
    retries = Retry(total=5, # nober of retries
                    backoff_factor=0.5, # growing time after each failure
                    status_forcelist=[429, 500, 502, 503, 504], # retries only on these HTTP errors
                    allowed_methods=["HEAD", "GET", "OPTIONS"]) # retry only for HTTP methods
    s.mount("https://", HTTPAdapter(max_retries=retries))
    s.mount("http://", HTTPAdapter(max_retries=retries))
    return s

session = create_session()
```

### 3) Checking robots.txt 

```python
# fetching the web content
url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population_(United_Nations)"
rp = robotparser.RobotFileParser()
rp.set_url(urljoin(url, "/robots.txt"))
rp.read()
can = rp.can_fetch("*", "/")

print("Allowed by roots?", can)
```
